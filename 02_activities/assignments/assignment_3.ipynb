#Assignment 3: Toronto Ferry Usage data 2015-2026 

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("Toronto Island Ferry Ticket Counts.csv")

# Convert timestamp
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Extract year
df["Year"] = df["Timestamp"].dt.year

# Sum yearly redemptions
yearly = (
    df.groupby("Year")["Redemption Count"]
      .sum()
      .reset_index()
)

# Create plot
plt.figure(figsize=(10,6))

plt.plot(
    yearly["Year"],
    yearly["Redemption Count"],
    marker="o"
)

plt.title(
    "Toronto Island Ferry Ticket Redemptions by Year (2015–2026)"
)

plt.xlabel("Year")
plt.ylabel("Total Tickets Redeemed")

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "annual_ferry_usage.png",
    dpi=300
)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("Toronto Island Ferry Ticket Counts.csv")

# Convert timestamp
df["Timestamp"] = pd.to_datetime(df["Timestamp"])

# Extract month
df["Month"] = df["Timestamp"].dt.month_name()

# Order months
month_order = [
    "January","February","March","April",
    "May","June","July","August",
    "September","October","November","December"
]

monthly = (
    df.groupby("Month")["Redemption Count"]
      .mean()
      .reindex(month_order)
)

# Plot
plt.figure(figsize=(10,6))

monthly.plot(kind="bar")

plt.title(
    "Average Ferry Ticket Redemptions by Month"
)

plt.xlabel("Month")
plt.ylabel("Average Redemptions per 15-Minute Interval")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "monthly_ferry_usage.png",
    dpi=300
)

plt.show()
```
