```python
# Monday June 1 Session Notes:
# Data visualization is dependent on:
# (1) Context - where and how it is used
# (2) Audience - who is interested
# (3) Data structure - what information does our data capture
#     (e.g. quantities, relationships)

# Making basic figures with matplotlib

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests

# Generate sample data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)

# Scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y)
plt.show()

# Bar plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x, y)
plt.show()

# Line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
plt.show()

# Histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
plt.show()

# Line plot with labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
plt.show()

# Font dictionaries
font1 = {
    'family': 'sans-serif',
    'color': 'blue',
    'size': 20
}

font2 = {
    'family': 'monospace',
    'color': 'green',
    'size': 14
}

# Line plot with custom fonts
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y)

ax.set_title(
    'Total growth over time',
    fontdict=font1,
    loc='left'
)

ax.set_ylabel(
    'Total growth',
    fontdict=font2
)

ax.set_xlabel(
    'Years since start',
    fontdict=font2
)

fig.tight_layout()
plt.show()

# Customized scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y, marker='*', color='indigo')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import requests

from PIL import Image
from matplotlib.ticker import (
    NullLocator,
    NullFormatter,
    MaxNLocator,
    MultipleLocator
)

# Reproducibility
np.random.seed(613)

x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)

# Basic line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)
plt.show()

# Legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='person1')
ax.plot(x, y2, label='person2')
ax.legend(loc='lower right', frameon=False)
plt.show()

# Customized legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='person1')
ax.plot(x, y2, label='person2')
ax.legend(
    loc='lower right',
    frameon=True,
    shadow=True,
    fontsize=12,
    ncol=2
)
plt.show()

# Legend outside plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='person1')
ax.plot(x, y2, label='person2')
ax.legend(loc='center', bbox_to_anchor=(1, 1))
plt.show()

# Legend positioned with anchor
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label='person1')
ax.plot(x, y2, label='person2')
ax.legend(loc='upper left', bbox_to_anchor=(0.5, 0.5))
plt.show()

# Text annotation
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label='person1')
ax.scatter(x, y2, label='person2')
ax.legend(loc='lower right')
ax.text(
    10,
    95,
    "This value is important!",
    ha='left',
    color='red',
    size=20
)
plt.show()

# Coordinate systems
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(
    1,
    5,
    ".Data: (1,5)",
    transform=ax.transData
)

ax.text(
    0.5,
    0.1,
    ".Axes: (0.5, 0.1)",
    transform=ax.transAxes
)

ax.text(
    0.2,
    0.2,
    ".Figure: (0.2, 0.2)",
    transform=fig.transFigure
)
plt.show()

# Annotation with arrow
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.annotate(
    'This is important!',
    xy=(10, 95),
    xytext=(20, 94),
    arrowprops=dict(facecolor='black')
)
plt.show()

# Custom arrow style
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.annotate(
    'This is important!',
    xy=(10, 95),
    xytext=(20, 94),
    arrowprops=dict(
        arrowstyle='wedge',
        color='hotpink'
    )
)
plt.show()

# Remove axis ticks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(NullLocator())
ax.xaxis.set_major_formatter(NullFormatter())
plt.show()

# Limit number of ticks
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(MaxNLocator(3))
plt.show()

# Set tick spacing
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(MultipleLocator(5))
plt.show()

# Rotate tick labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
plt.show()

# Custom fonts
font1 = {
    'family': 'serif',
    'color': 'indigo'
}

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="person 1")
ax.scatter(x, y2, label="person 2")
ax.legend(loc='lower right')

plt.xlabel(
    'Shiny New X Axis!',
    fontsize=18,
    fontdict=font1
)
plt.show()

# Available styles
print(plt.style.available)

# Apply style
plt.style.use('fivethirtyeight')

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1)
ax.plot(x, y2)

plt.show()
```

```python
# Wednesday June 10 Visualization with Purpose - Accessible Data Visualization

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests

# ------------------------------------------------------------------
# Sample Data
# ------------------------------------------------------------------

np.random.seed(613)

x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)

x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([100, 180, 240, 99, 220])

# ------------------------------------------------------------------
# Scatterplot and Bar Chart
# ------------------------------------------------------------------

fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

ax1.scatter(x1, y1)
ax1.set_title("Scatterplot")

ax2.bar(x2, y2)
ax2.set_title("Bar Chart")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------
# Scatterplot with Annotation
# ------------------------------------------------------------------

fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

ax1.scatter(x1, y1)

ax1.annotate(
    "This is important",
    xy=(10, y1[10]),
    xytext=(25, 70),
    arrowprops=dict(facecolor="black")
)

ax2.bar(x2, y2)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------
# Mosaic Layout
# ------------------------------------------------------------------

fig, someaxes = plt.subplot_mosaic(
    [
        ["ax1", "ax3"],
        ["ax2", "ax3"]
    ],
    figsize=(7, 4)
)

someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel("A Big Label", fontsize=18)
someaxes["ax2"].set_xlabel("Another Label", fontsize=18)
someaxes["ax3"].set_xlabel("Label 2: 2 Fast 2 Furious", fontsize=18)

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------
# Bar Chart with Line Plot
# ------------------------------------------------------------------

x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])

y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7, 3))

ax.bar(x, y1, color="indigo")
ax.plot(x, y2, color="red", marker="o")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------------
# Line Plot with Error Bars
# ------------------------------------------------------------------

y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7, 3))

ax.plot(x, y2, color="red", marker="o")

ax.errorbar(
    x,
    y2,
    yerr=y2_sd,
    fmt="none",
    ecolor="indigo",
    elinewidth=4,
    capsize=6,
    capthick=4
)

plt.tight_layout()
plt.show()
```

```python
#URL Images Upload 

from PIL import Image
import requests
from io import BytesIO
import matplotlib.pyplot as plt
import numpy as np

# Example data
x = np.arange(50)
y2 = np.random.randint(0, 75, 50)

# Image URL (replace with the actual image URL)
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response = requests.get(url, headers=headers)

image_file = BytesIO(response.content)
image = Image.open(image_file)

# Create figure and plot
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color="red")

# Add image to figure
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])
ax_image.imshow(image)
ax_image.axis("off")

plt.show()

# Save figure (optional)
# path = "./"
# filename = "plot_with_image.png"
# plt.savefig(path + filename, dpi=300, bbox_inches="tight")
```

```python
# June 15 - Data Visualization as Advocacy & Beyond MatPlotLib
# Seaborn is a data viz package built on top of matplotlib.
# Similar, but seaborn takes care of semantic mapping in addition.

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests

# Load and inspect dataset
tips = sns.load_dataset("tips")
print(tips)

# Set plotting style
sns.set_style('whitegrid')

# Basic line plot
tipgraph = sns.lineplot(
    data=tips,
    x='total_bill',
    y='tip'
)

tipgraph.set(
    title='Tips vs. Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)'
)

plt.show()

# Customized line plot
fig, ax = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(
    data=tips,
    x='total_bill',
    y='tip',
    color='hotpink',
    linestyle='--',
    linewidth=3,
    marker='o',
    markerfacecolor='indigo'
)

tipgraph.set(
    title='Tips vs. Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)'
)

plt.show()

# Scatter plot with semantic mapping
fig, ax = plt.subplots(figsize=(8, 5))

tipgraph = sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    style='time',
    hue='day',
    palette=['purple', 'hotpink', 'deepskyblue', 'yellowgreen']
)

tipgraph.set(
    title='Tips vs. Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)'
)

plt.show()

# Pairplot is another convenient way to compare variables
sns.pairplot(data=tips, hue='day')
plt.show()

# Relplot lets us explore specific levels within variables
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter",
    col_wrap=2
)

plt.show()

# Plotly
import plotly.graph_objects as go  # 'go' is 'graph objects'

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

# Plotly syntax is different from matplotlib/seaborn
graph = go.Figure()

graph.add_trace(
    go.Bar(
        x=x1,
        y=y1
    )
)

graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score"
)

graph.show()

# Save interactive graph
graph.write_html("pirategraph.html")

# Interactive scatter plot
graph = go.Figure()

graph.add_trace(
    go.Scatter(
        x=x1,
        y=y1,
        mode='markers',
        marker=dict(
            size=15,
            color='hotpink',
            opacity=1,
            line=dict(
                width=5,
                color='purple'
            )
        )
    )
)

graph.update_layout(
    title='Interactive Pirate Plot',
    xaxis_title='Pirates',
    yaxis_title='Scores',
    width=500,
    height=500
)

graph.show()

# Making word clouds
from wordcloud import WordCloud

df = pd.read_csv(
    "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
    on_bad_lines='skip'
)

print(df.head())

# Join all text from the quote column into one string
text = " ".join(str(each) for each in df.quote)

# Generate word cloud
wordcloud = WordCloud(
    background_color="white",
    colormap='inferno'
).generate(text)

# Display word cloud
fig, ax = plt.subplots(figsize=(7, 3))

ax.imshow(
    wordcloud,
    interpolation='bilinear'
)

ax.axis("off")
plt.show()

# Making Venn diagrams
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = {"apple", "banana", "watermelon"}
B = {"pumpkin", "blueberry", "apple", "key lime"}

diagram = venn2_unweighted(
    [A, B],
    set_labels=('Fruits', 'Pies'),
    set_colors=("blue", "red"),
    alpha=0.5
)

diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.show()
```
